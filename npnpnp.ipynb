```python
'''
1)массив покупок в рублях:[150,80,320,45,200,90]
выведите только покупки дороже 100
'''
ps=np.array([150,80,320,45,200,90])
print(ps[ps>100],'\n')


'''
2)оценки:[2,3,5,4,2,5]
замените:если оценка >=4->'зачёт',иначе->'незачёт'
'''
оценки=np.array([2,3,5,4,2,5],)
print(np.where(оценки>=4,'зачёт','незачёт'),'\n')


'''
3)1.массив целых от 5 до 14 включительно
2.ровно 5 чисел 0 до 1 включительно
'''
print(f'1.{np.arange(start=5,stop=15)}','\n')
print(f'2.{np.linspace(0,1,5)}','\n')


'''
4)создайте:
вектор и 5 нулей
табицу 2*3 из единиц
матрицу 3*3 с единицами по диагонали
'''
print(np.zeros(5),'\n')
print(np.ones([2,3]),'\n')
print(np.eye(3),'\n')


'''
5)выведите:элeмент 50;первую строку;второй столбец
'''
mx=np.array([(10,20,30),(40,50,60),(70,80,90)])
print(mx)
print(mx[1,1])
print(mx[0,:])
print(mx[:,1],'\n')


'''
6)посчитайте среднее,стандартное отклонение,медиану,минимум,максимум
'''
mix=np.array([10,12,9,15,11,8])
print(mix)
print(mix.mean())
print(mix.std())
print(np.median(mix))
print(mix.min())
print(mix.max(),'\n')
```
