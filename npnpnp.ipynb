преобразовать массив весов из кг в тонны

Пусть необходимо выбрать только элементы массива со значением веса меньше 100 кг (0.1 тонна) - поделите массив на две части

```python
weight_kg = [500, 1100, 1800, 81.65, 97.52, 95.25, 92.98, 86.18, 88.45, 100., 120.,240.,]

# Create a numpy array np_weight_kg from weight_kg
np_weight_kg=np.array(weight_kg)
# Create np_weight_ton from np_weight_kg
np_weight_ton=np_weight_kg/1000
# Print out np_weight_ton
print(np_weight_ton)
```

```python
'''
массив покупок в рублях:[150,80,320,45,200,90]
выведите только покупки дороже 100
'''
ps=np.array([150,80,320,45,200,90])
print(ps[ps>100],'\n')

'''
оценки:[2,3,5,4,2,5]
замените:если оценка >=4->'зачёт',иначе->'незачёт'
'''

оценки=np.array([2,3,5,4,2,5],)
print(np.where(оценки>=4,'зачёт','незачёт'),'\n')

'''
1.массив целых от 5 до 14 включительно
2.ровно 5 чисел 0 до 1 включительно
'''

print(f'1.{np.arange(start=5,stop=15)}','\n')
print(f'2.{np.linspace(0,1,5)}','\n')

'''
создайте:
вектор и 5 нулей
табицу 2*3 из единиц
матрицу 3*3 с единицами по диагонали
'''

print(np.zeros(5),'\n')
print(np.ones([2,3]),'\n')
print(np.eye(3))
```
